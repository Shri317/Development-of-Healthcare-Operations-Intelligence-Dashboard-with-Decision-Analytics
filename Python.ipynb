## Question 1: Check Even or Odd (Basic Programming)

**Problem Statement:** Write a Python function to check if a given number is even or odd.

**Example Input:**
```python
n = 4
```
**Expected Output:**
```
Even
```

````python
def check_even_odd(n):
    if n % 2 == 0:
        return "Even"
    else:
        return "Odd"

# Example usage:
print(check_even_odd(4))
print(check_even_odd(7))
````

**Short Explanation:** The modulo operator (`%`) returns the remainder of a division. If `n % 2` is 0, the number is even; otherwise, it's odd.

## Question 2: Calculate Factorial (Numbers)

**Problem Statement:** Write a Python function to calculate the factorial of a non-negative integer.

**Example Input:**
```python
n = 5
```
**Expected Output:**
```
120
```

````python
def factorial(n):
    if n == 0:
        return 1
    else:
        res = 1
        for i in range(1, n + 1):
            res *= i
        return res

# Example usage:
print(factorial(5))
print(factorial(0))
````

**Short Explanation:** The function uses an iterative approach. If `n` is 0, the factorial is 1. Otherwise, it multiplies all numbers from 1 to `n`.

## Question 3: Reverse a String (Strings)

**Problem Statement:** Write a Python function to reverse a given string.

**Example Input:**
```python
s = "hello"
```
**Expected Output:**
```
olleh
```

````python
def reverse_string(s):
    return s[::-1]

# Example usage:
print(reverse_string("hello"))
print(reverse_string("Python"))
````

**Short Explanation:** Python's string slicing provides a concise way to reverse a string. `[::-1]` creates a reversed copy of the string.

## Question 4: Find Largest Element in a List (Lists and Arrays)

**Problem Statement:** Write a Python function to find the largest element in a list of numbers. Assume the list is not empty.

**Example Input:**
```python
numbers = [3, 1, 4, 1, 5, 9, 2]
```

## Question 5: Count Word Frequency (Dictionaries and Sets)

**Problem Statement:** Write a Python function to count the frequency of each word in a given sentence. The words should be case-insensitive.

**Example Input:**
```python
sentence = "Hello world hello Python"
```
**Expected Output:**
```python
{'hello': 2, 'world': 1, 'python': 1}
```

````python
def count_word_frequency(sentence):
    words = sentence.lower().split()
    word_counts = {}
    for word in words:
        word_counts[word] = word_counts.get(word, 0) + 1
    return word_counts

# Example usage:
print(count_word_frequency("Hello world hello Python"))
print(count_word_frequency("Apple banana Apple"))
````

**Short Explanation:** The sentence is converted to lowercase and split into words. A dictionary stores word counts, incrementing a word's count each time it appears. `get(word, 0)` ensures a new word starts with a count of 0.

## Question 6: Check for Palindrome (Strings)

**Problem Statement:** Given a string `s`, return `True` if it's a palindrome, `False` otherwise. Ignore case.

**Example Input:**
```python
s = "madam"
```
**Expected Output:**
```
True
```

````python
def is_palindrome(s):
    s = s.lower()
    return s == s[::-1]

# Example usage:
print(is_palindrome("madam"))
print(is_palindrome("racecar"))
print(is_palindrome("hello"))
print(is_palindrome("Madam"))
````

**Short Explanation:** The function first converts the string to lowercase to ignore case. Then, it compares the original string with its reversed version using string slicing.

## Question 8: Find Common Elements Between Two Lists (Lists and Arrays)

**Problem Statement:** Given two lists, `list1` and `list2`, return a new list containing only the elements that are present in both lists. The order of elements in the output list does not matter.

**Example Input:**
```python
list1 = [1, 2, 3, 4]
list2 = [3, 4, 5, 6]
```
**Expected Output:**
```python
[3, 4] # Order might vary
```

````python
def find_common_elements(list1, list2):
    return list(set(list1) & set(list2))

# Example usage:
print(find_common_elements([1, 2, 3, 4], [3, 4, 5, 6]))
print(find_common_elements(['apple', 'banana', 'orange'], ['orange', 'grape', 'banana']))
````

**Short Explanation:** Convert both lists to sets. The intersection (`&`) of two sets gives common elements. Convert the resulting set back to a list.

## Question 9: Check if Key Exists in Dictionary (Dictionaries and Sets)

**Problem Statement:** Given a dictionary `d` and a `key`, return `True` if the `key` is present in `d`, otherwise return `False`.

**Example Input:**
```python
my_dict = {'a': 1, 'b': 2, 'c': 3}
key_to_check = 'b'
```
**Expected Output:**
```
True
```

````python
def check_key_exists(d, key):
    return key in d

# Example usage:
my_dict = {'a': 1, 'b': 2, 'c': 3}
print(check_key_exists(my_dict, 'b'))
print(check_key_exists(my_dict, 'd'))
````

**Short Explanation:** The `in` operator is the most Pythonic way to check for the presence of a key in a dictionary.

## Question 10: Remove Duplicates from a List (Lists and Arrays / Sets)

**Problem Statement:** Given a list `input_list`, return a new list containing only the unique elements from the original list. The order of elements in the output list does not need to be preserved.

**Example Input:**
```python
input_list = [1, 2, 2, 3, 4, 4, 5]
```
**Expected Output:**
```python
[1, 2, 3, 4, 5] # Order might vary
```

````python
def remove_duplicates(input_list):
    return list(set(input_list))

# Example usage:
print(remove_duplicates([1, 2, 2, 3, 4, 4, 5]))
print(remove_duplicates(['apple', 'banana', 'apple', 'orange']))
````

**Short Explanation:** Converting the list to a set automatically removes duplicate elements, as sets only store unique values. Converting the set back to a list gives the desired output.
