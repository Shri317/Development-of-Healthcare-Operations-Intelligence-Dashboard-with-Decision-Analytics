# SQL Coding Interview Questions for Freshers

To begin, we'll set up an in-memory SQLite database using `sqlite3` in Python. This will allow us to create tables and insert data directly within the Colab environment, and then run SQL queries against it.

````python
import sqlite3
import pandas as pd

# Connect to an in-memory SQLite database
conn = sqlite3.connect(':memory:')
c = conn.cursor()

# Create Employees table
c.execute('''
CREATE TABLE Employees (
    EmpID INTEGER PRIMARY KEY,
    EmpName TEXT NOT NULL,
    DepartmentID INTEGER,
    Salary INTEGER
);
''')

# Create Departments table
c.execute('''
CREATE TABLE Departments (
    DepartmentID INTEGER PRIMARY KEY,
    DepartmentName TEXT NOT NULL
);
''')

# Insert data into Employees
c.execute("INSERT INTO Employees (EmpID, EmpName, DepartmentID, Salary) VALUES (1, 'Alice', 101, 60000);")
c.execute("INSERT INTO Employees (EmpID, EmpName, DepartmentID, Salary) VALUES (2, 'Bob', 102, 80000);")
c.execute("INSERT INTO Employees (EmpID, EmpName, DepartmentID, Salary) VALUES (3, 'Charlie', 101, 65000);")
c.execute("INSERT INTO Employees (EmpID, EmpName, DepartmentID, Salary) VALUES (4, 'David', 102, 90000);")
c.execute("INSERT INTO Employees (EmpID, EmpName, DepartmentID, Salary) VALUES (5, 'Eve', 101, 70000);")
c.execute("INSERT INTO Employees (EmpID, EmpName, DepartmentID, Salary) VALUES (6, 'Frank', 103, 75000);")

# Insert data into Departments
c.execute("INSERT INTO Departments (DepartmentID, DepartmentName) VALUES (101, 'HR');")
c.execute("INSERT INTO Departments (DepartmentID, DepartmentName) VALUES (102, 'Engineering');")
c.execute("INSERT INTO Departments (DepartmentID, DepartmentName) VALUES (103, 'Marketing');")

# Commit the changes
conn.commit()

print("Database and tables created with sample data.")

# Helper function to run SQL queries and display results in a DataFrame
def run_sql_query(query):
    df = pd.read_sql_query(query, conn)
    return df
````

### Question 1: Select All Data from a Table

**Problem Statement:** Retrieve all columns and all rows from the `Employees` table.

**Sample Table: `Employees`**
```

**Expected Output:**
```

````python
# SQL Query for Question 1
query1 = """
SELECT *
FROM Employees;
"""
display(run_sql_query(query1))

# Explanation:
# The `SELECT *` statement is used to retrieve all columns from a table.
# The `FROM Employees` specifies that the data should be retrieved from the `Employees` table.
````

### Question 6: Count Records in a Table

**Problem Statement:** Count the total number of employees in the `Employees` table.

**Sample Table: `Employees`**
```
EmpID | EmpName    | DepartmentID | Salary
-------|------------|--------------|--------
1     | Alice      | 101          | 60000
2     | Bob        | 102          | 80000
3     | Charlie    | 101          | 65000
4     | David      | 102          | 90000
```

**Expected Output:**
```
TotalEmployees
---------------
6
```

````python
# SQL Query for Question 6
query6 = """
SELECT COUNT(EmpID) AS TotalEmployees
FROM Employees;
"""
display(run_sql_query(query6))

# Explanation:
# The `COUNT()` aggregate function is used to count the number of rows that match a specified criterion. Using `COUNT(column_name)` counts non-NULL values in that column.
# `AS TotalEmployees` renames the output column for clarity.
````

### Question 7: Group By and Aggregate Functions (SUM)

**Problem Statement:** Calculate the total `Salary` for each `DepartmentID`.

**Sample Table: `Employees`**
```
EmpID | EmpName    | DepartmentID | Salary
-------|------------|--------------|--------
1     | Alice      | 101          | 60000
2     | Bob        | 102          | 80000
3     | Charlie    | 101          | 65000
4     | David      | 102          | 90000
5     | Eve        | 101          | 70000
```

**Expected Output:**
```
DepartmentID | TotalSalary
--------------|-------------
101          | 195000
102          | 170000
103          | 75000
```

````python
# SQL Query for Question 7
query7 = """
SELECT DepartmentID, SUM(Salary) AS TotalSalary
FROM Employees
GROUP BY DepartmentID;
"""
display(run_sql_query(query7))

# Explanation:
# The `SUM()` aggregate function calculates the sum of a numeric column.
# The `GROUP BY` clause groups rows that have the same values in specified columns into summary rows, like "find the sum of salary for each department ID".
````

### Question 8: Group By and Filter with HAVING Clause

**Problem Statement:** Find department IDs where the total `Salary` is greater than 150,000.

**Sample Table: `Employees`**
```
EmpID | EmpName    | DepartmentID | Salary
-------|------------|--------------|--------
1     | Alice      | 101          | 60000
2     | Bob        | 102          | 80000
3     | Charlie    | 101          | 65000
4     | David      | 102          | 90000
5     | Eve        | 101          | 70000
```

**Expected Output:**
```
DepartmentID | TotalSalary
--------------|-------------
101          | 195000
102          | 170000
```

````python
# SQL Query for Question 8
query8 = """
SELECT DepartmentID, SUM(Salary) AS TotalSalary
FROM Employees
GROUP BY DepartmentID
HAVING SUM(Salary) > 150000;
"""
display(run_sql_query(query8))

# Explanation:
# `HAVING` is used to filter records based on aggregate functions, unlike `WHERE` which filters individual rows.
# Here, it filters the grouped departments, keeping only those where the `TotalSalary` is greater than 150,000.
````

### Question 9: INNER JOIN Two Tables

**Problem Statement:** List all employees along with their respective department names.

**Sample Table 1: `Employees`**
```
EmpID | EmpName    | DepartmentID | Salary
-------|------------|--------------|--------
1     | Alice      | 101          | 60000
2     | Bob        | 102          | 80000
```

**Sample Table 2: `Departments`**
```
DepartmentID | DepartmentName
--------------|---------------
101          | HR
102          | Engineering
```

**Expected Output:**
```
EmpName    | DepartmentName
------------|-----------------
Alice      | HR
Bob        | Engineering
Charlie    | HR
David      | Engineering
Eve        | HR
Frank      | Marketing
```

````python
# SQL Query for Question 9
query9 = """
SELECT E.EmpName, D.DepartmentName
FROM Employees AS E
INNER JOIN Departments AS D ON E.DepartmentID = D.DepartmentID;
"""
display(run_sql_query(query9))

# Explanation:
# An `INNER JOIN` returns rows when there is a match in both tables based on a specified condition.
# The `ON E.DepartmentID = D.DepartmentID` specifies the common column to join the tables.
# Aliases `E` and `D` are used for brevity to refer to the `Employees` and `Departments` tables, respectively.
````

### Question 10: Subquery to Find Employees with High Salary

**Problem Statement:** Find the names of employees whose `Salary` is greater than the average `Salary` of all employees.

**Sample Table: `Employees`**
```
EmpID | EmpName    | DepartmentID | Salary
-------|------------|--------------|--------
1     | Alice      | 101          | 60000
2     | Bob        | 102          | 80000
3     | Charlie    | 101          | 65000
4     | David      | 102          | 90000
5     | Eve        | 101          | 70000
```

**Expected Output:**
```
EmpName
------------
Bob
David
Frank
```
(Note: Average Salary = (60000+80000+65000+90000+70000+75000)/6 = 73333.33)

````python
# SQL Query for Question 10
query10 = """
SELECT EmpName
FROM Employees
WHERE Salary > (SELECT AVG(Salary) FROM Employees);
"""
display(run_sql_query(query10))

# Explanation:
# A subquery (the inner `SELECT AVG(Salary) FROM Employees`) executes first and returns a single value (the average salary).
# The outer query then uses this value to filter employees whose individual `Salary` is greater than the calculated average.
````
